e) Calculate the following statistics:


```python
# i. The average salary of employees
average_salary = df_employees['Salary'].mean()
print(f"i. The average salary of employees: ${average_salary:,.2f}")

# ii. The total bonus paid to all employees
total_bonus = df_employees['Bonus'].sum()
print(f"ii. The total bonus paid to all employees: ${total_bonus:,.2f}")

# iii. The youngest employee's age
youngest_age = df_employees['Age'].min()
print(f"iii. The youngest employee's age: {youngest_age} years")

# iv. The highest performance rating
highest_rating = df_employees['Rating'].max()
print(f"iv. The highest performance rating: {highest_rating}")
```

### f) Sort the DataFrame by the Salary column in descending order.

```python
df_sorted_by_salary = df_employees.sort_values(by='Salary', ascending=False)
print("DataFrame sorted by Salary in descending order:")
display(df_sorted_by_salary)
```

### g) Add a new column that categorizes employees based on their performance rating:
i. Excellent for ratings >= 4.5
ii. Good for ratings >= 4.0 but < 4.5
iii. Average for ratings < 4.0

```python
def categorize_rating(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0 and rating < 4.5:
        return 'Good'
    else:
        return 'Average'

df_employees['Performance_Category'] = df_employees['Rating'].apply(categorize_rating)
print("DataFrame with 'Performance_Category' column:")
display(df_employees)
```

### h) Identify missing values in the DataFrame.

```python
print("Missing values in the DataFrame:")
print(df_employees.isnull().sum())
```

### i) Rename the Employee_ID column to ID.

```python
df_employees.rename(columns={'Employee_ID': 'ID'}, inplace=True)
print("DataFrame after renaming 'Employee_ID' to 'ID':")
display(df_employees)
```

### j) Find all employees who:
i. Have more than 5 years of experience.
ii. Belong to the IT department.

```python
employees_more_than_5_years_exp = df_employees[df_employees['Years_of_Experience'] > 5]
print("Employees with more than 5 years of experience:")
display(employees_more_than_5_years_exp)
```

```python
employees_it_department = df_employees[df_employees['Department'] == 'IT']
print("Employees belonging to the IT department:")
display(employees_it_department)
```

### k) Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary.

```python
df_employees['Tax'] = df_employees['Salary'] * 0.10
print("DataFrame with new 'Tax' column:")
display(df_employees)
```

```python
new_csv_file_path = 'modified_employees.csv'
df_employees.to_csv(new_csv_file_path, index=False)
print(f"Modified DataFrame saved successfully to '{new_csv_file_path}'.")
```
