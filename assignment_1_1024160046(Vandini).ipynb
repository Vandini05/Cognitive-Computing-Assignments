sum of all prime numbers entered till the entered number

```python
n = int(input("Enter n: "))
total_sum = 0

for i in range(2, n + 1):
    if all(i % j != 0 for j in range(2, i)):
        total_sum += i

print("Sum of primes:", total_sum)
```

Sum of odd numbers till the entered numbers

```python
def add_odd(n):
    k=0
    for i in range(1,n+1,1):
      if i%2!=0:
        k=k+i
    return k
add_odd(88)
```

sum of prime numbers till the entered number

```python
def add_primes(n):
    total_sum = 0
    for i in range(2, n + 1):
        is_prime = True
        for j in range(2, int(i**0.5) + 1):
            if i % j == 0:
                is_prime = False
                break
        if is_prime:
            total_sum += i
    return total_sum

n_input = int(input("Enter a number (n): "))
sum_of_primes = add_primes(n_input)
print(f"Sum of prime numbers up to {n_input} is: {sum_of_primes}")
```
