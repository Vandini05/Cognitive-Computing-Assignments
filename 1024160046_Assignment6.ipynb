```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print(f"Original temperature readings: {temperature}")
```

### a. Add 2°C to every reading using vectorization.

```python

corrected_temperature = temperature + 2
print(f"Temperature readings after adding 2°C: {corrected_temperature}")
```

### b. Convert all temperatures from Celsius to Fahrenheit.

```python

temperature_fahrenheit = (9/5) * corrected_temperature + 32
print(f"Temperature readings in Fahrenheit: {temperature_fahrenheit}")
```

### c. Identify all readings greater than 32°C using Boolean indexing.

```python

higher_than_32_celsius_mask = corrected_temperature > 32
readings_greater_than_32c = corrected_temperature[higher_than_32_celsius_mask]
print(f"Readings greater than 32°C: {readings_greater_than_32c}")
```

### d. Count how many readings exceed 32°C.

```python

count_greater_than_32c = np.sum(higher_than_32_celsius_mask)
print(f"Number of readings exceeding 32°C: {count_greater_than_32c}")
```

### e. Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value.

**Vectorization** is the process of performing operations on entire arrays or collections of data at once, rather than iterating through individual elements. In NumPy, this is highly optimized because operations are implemented in C and Fortran, allowing them to leverage CPU-level optimizations (like SIMD instructions). This leads to significantly faster execution times, especially for large datasets, compared to Python's native for loops, which incur overhead for each iteration.

**Boolean indexing** is a powerful feature where you use a Boolean array (an array of True/False values) to select elements from another array. Instead of writing conditional statements within a loop to check each element, Boolean indexing allows you to filter data concisely and efficiently. Like vectorization, it's implemented at a lower level, providing performance benefits over explicit iteration and conditional checks in Python loops. It's also more readable and less prone to off-by-one errors.

## Q2. Operations on a 2D NumPy Array (Steps Data)

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
print(f"Original steps matrix:\n{steps}")
```

### a. Calculate the total steps recorded.

```python
total_steps = np.sum(steps)
print(f"Total steps recorded: {total_steps}")
```

### b. Calculate the mean number of steps.

```python
mean_steps = np.mean(steps)
print(f"Mean number of steps: {mean_steps:.2f}")
```

### c. Find the maximum and minimum values.

```python
max_steps = np.max(steps)
min_steps = np.min(steps)
print(f"Maximum steps recorded: {max_steps}")
print(f"Minimum steps recorded: {min_steps}")
```

### d. Calculate the total steps for each day using `axis=0`.

```python
total_steps_per_day = np.sum(steps, axis=0)
print(f"Total steps for each day: {total_steps_per_day}")
print(f"  (Column 0: Day 1, Column 1: Day 2, Column 2: Day 3)")
```

### e. Calculate the total steps for each user using `axis=1`.

```python
total_steps_per_user = np.sum(steps, axis=1)
print(f"Total steps for each user: {total_steps_per_user}")
print(f"  (Row 0: User 1, Row 1: User 2, etc.)")
```

### f. Find the user/day position containing the maximum number of steps using `argmax()`.

```python
max_steps_flat_index = np.argmax(steps)
max_steps_2d_coords = np.unravel_index(max_steps_flat_index, steps.shape)

user_index = max_steps_2d_coords[0]
day_index = max_steps_2d_coords[1]

print(f"Position of maximum steps (user, day): {max_steps_2d_coords}")
print(f"This corresponds to User {user_index + 1} on Day {day_index + 1} with {steps[user_index, day_index]} steps.")
```

## Q4. Ordinary Least Squares (OLS) with NumPy

```python
import numpy as np

y = np.array([40, 65, 30, 85])
print(f"Dependent variable (y): {y}")

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
print(f"\nIndependent variables (X):\n{X}")
```

### a. Display the shape and dimensions of X.

```python
print(f"Shape of X: {X.shape}")
print(f"Dimensions of X (ndim): {X.ndim}")
```

### b. Calculate `X.T` and explain what the transpose represents.

```python
X_transpose = X.T
print(f"X transpose (X.T):\n{X_transpose}")
print("\nExplanation: The transpose of a matrix (X.T) is obtained by flipping the matrix over its diagonal. Rows become columns, and columns become rows. In this context, the original matrix X had rows representing users and columns representing features (sleep hours, activity level, stress level). The transposed matrix X.T now has rows representing the features and columns representing the users. This operation is crucial in linear algebra, especially for matrix multiplication in OLS, to align the dimensions correctly.")
```

### c. Calculate the matrix product: `X.T @ X`

```python
XT_X = X_transpose @ X
print(f"Matrix product (X.T @ X):\n{XT_X}")
```

### d. Calculate the matrix inverse using: `np.linalg.inv()` where appropriate.

```python
XT_X_inverse = np.linalg.inv(XT_X)
print(f"Matrix inverse of (X.T @ X):\n{XT_X_inverse}")
```

### e. Implement the Ordinary Least Squares equation: $\beta = (X^T X)^{-1} X^T y$ using NumPy.

```python

XT_y = X_transpose @ y
print(f"Matrix product (X.T @ y): {XT_y}")

beta_coefficients = XT_X_inverse @ XT_y
print(f"\nOLS Beta Coefficients: {beta_coefficients}")
```

### f. Explain what the resulting coefficients represent in the context of the three user features.

**Explanation of Coefficients:**

The beta_coefficients calculated are the estimated parameters for our linear regression model. In the context of the three user features (Sleep hours, Activity level, Stress level) and the predicted Assistance Score, these coefficients represent the following:

*   **beta_coefficients[0] (for Sleep hours):** This coefficient indicates the expected change in the assistance score for every one-unit increase in 'Sleep hours', assuming 'Activity level' and 'Stress level' remain constant. A positive value would suggest that more sleep hours are associated with a higher assistance score, while a negative value would suggest the opposite.

*   **beta_coefficients[1] (for Activity level):** This coefficient indicates the expected change in the assistance score for every one-unit increase in 'Activity level', assuming 'Sleep hours' and 'Stress level' remain constant. For example, a positive value means higher activity is associated with a higher assistance score.

*   **beta_coefficients[2] (for Stress level):** This coefficient indicates the expected change in the assistance score for every one-unit increase in 'Stress level', assuming 'Sleep hours' and 'Activity level' remain constant. A negative value here would typically suggest that higher stress levels are associated with a lower assistance score.

In essence, these coefficients quantify the independent impact of each feature on the assistance score within this linear model.

### g. A new user has: `new_user = np.array([5, 40, 7])`. Use your calculated coefficients to obtain a predicted assistance score.

```python
new_user = np.array([5, 40, 7])
print(f"New user features: {new_user}")


predicted_assistance_score = new_user @ beta_coefficients

print(f"\nPredicted assistance score for the new user: {predicted_assistance_score:.2f}")
```
